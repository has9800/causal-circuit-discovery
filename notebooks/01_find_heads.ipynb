```python
from pathlib import Path
import sys
import yaml

ROOT = Path('..').resolve()
if str(ROOT) not in sys.path:
    sys.path.insert(0, str(ROOT))

from src.model import load_model
from src.record import load_pairs, record_pair_activations
from src.compare import compare_conditions, top_heads
from src.ablate import evaluate_ablation
from src.visualize import save_all_visuals
```

```python
cfg = yaml.safe_load((ROOT / 'config.yaml').read_text())
model = load_model(cfg['model'])
pairs = load_pairs(str(ROOT / cfg['data']['pairs_path']))
```

```python
record = record_pair_activations(model, pairs)
comparison = compare_conditions(record, cfg['analysis']['top_k'], cfg['analysis']['significance_threshold'])
heads = top_heads(comparison.rank_indices, cfg['analysis']['ablation_k'])
ablation_results = evaluate_ablation(model, pairs, heads, cfg['analysis']['ablation_k'], cfg['analysis']['n_random_baselines'])
save_all_visuals(comparison, ablation_results, str(ROOT / cfg['output']['dir']))
ablation_results
```
